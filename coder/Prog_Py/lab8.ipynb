```python
import numpy as np
import matplotlib.pyplot as plt

days = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс'] * 2
day_week = np.array([5.1, 6.3, 4.8, 7.2, 9.5, 8.1, 6.7, 5.9, 7.4, 8.8, 10.2, 9.1, 7.5, 6.2])
night_week = np.array([-1.2, 0.3, -2.1, 1.5, 3.8, 2.4, 0.9, -0.4, 1.8, 3.2, 4.6, 2.9, 1.1, -0.3])

x = np.arange(len(days))
mean_night = night_week.mean()
mean_day = day_week.mean()
fig, ax = plt.subplots(figsize=(11, 4.5))

ax.plot(x, day_week, marker='o', color='orange', linewidth=2, markersize=6, label='Темп. днем')
ax.plot(x, night_week, marker='s', color='blue', linewidth=2, markersize=6, label='Темп. ночью')

ax.axhline(mean_day, color='orange', linestyle='--', linewidth=1, alpha=0.7, label=f'Ср. днем = {mean_day:.1f}°C')
ax.axhline(mean_night, color='blue', linestyle='--', linewidth=1, alpha=0.7, label=f'Ср. ночью = {mean_night:.1f}°C')
ax.axhline(0, color='red', linestyle='-', linewidth=1.5, alpha=0.8, label='Граница мороза (0°C)')

plt.title('Температура за 2 недели')

ax.set_xticks(x)
ax.set_xticklabels(days, rotation=45)

ax.set_xlabel('День недели')
ax.set_ylabel('Температура, °C')
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc='upper left')

fig.tight_layout()
fig.savefig('weather.png', dpi=150)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

days = list(range(1, 15))
screen = np.array([4.5, 5.2, 5.8, 4.9, 6.1, 6.8, 7.2, 7.5, 8.1, 7.8, 8.5, 9.0, 9.3, 9.7])
battery = np.array([58, 52, 48, 55, 42, 38, 33, 29, 22, 25, 18, 14, 11, 8])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.plot(days, screen, marker='o', color='blue', label='Экран, ч')
ax2.plot(days, battery, marker='o', color='green', label='Батарея, %')

mean_screen = sum(screen) / len(screen)
mean_battery = sum(battery) / len(battery)

ax1.axhline(mean_screen, color='blue', linestyle='--', alpha=0.6, label=f'Среднее: {mean_screen:.1f} ч')
ax2.axhline(mean_battery, color='green', linestyle='--', alpha=0.6, label=f'Среднее: {mean_battery:.1f}%')

ax1.set_ylim(0, 11)
ax2.set_ylim(0, 100)

ax1.set_title('Эрканное время')
ax1.set_xlabel('День')
ax1.set_ylabel('Часы')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend()

ax2.set_title('Заряд батареи')
ax2.set_xlabel('День')
ax2.set_ylabel('Проценты (%)')
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.savefig('phone.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

names = ['Маргарита Миния', 'Пепперони-Парад', 'Гавайский Хаос', 'Четыре Сыра и Слезы', 'Веганский Бунт']
sales = [23, 45, 56, 78, 32]

x = np.arange(len(names))
mx = np.argmax(sales)

fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(names, sales, width=0.6, color='pink', edgecolor='red')

ax.set_title('Битва пиццерий: продажи за неделю')

ax.set_xlabel('Пиццерия')
ax.set_ylabel('Продано пицц, шт.')

plt.tight_layout()
plt.savefig('pizzeria.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ['спасает мир', 'восстанавливает суперсилы', 'отдыхает', 'Занимается <<другим>>']
hours = [8, 7, 5, 4]

fig, ax = plt.subplots(figsize=(7, 7))

ax.pie(hours, labels=labels, colors=('red', 'blue', 'green', 'yellow'), autopct='%1.1f%%')
ax.set_title('День супергероя Алексея')

plt.tight_layout()
plt.savefig('superhero.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

c, n = 300000, 1.5
v = c / n
dist = np.array([100, 500, 1000, 3000, 5000, 10000])
ping = np.array([1.8, 5.9, 11.5, 32.5, 53.0, 104.2])
teor_ping = (2 * dist / v) * 1000

plt.figure(figsize=(10, 6))

plt.plot(dist, ping, linestyle='None', marker='o', color='blue', label='Экспериментальный пинг')
plt.plot(dist, teor_ping, linestyle='--', color='red', label='Теоретический пинг')

plt.title('Зависимость пинга от расстояния')
plt.xlabel('Расстояние, км')
plt.ylabel('Пинг, мс')

plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

plt.tight_layout()

plt.savefig('ping_test.png', dpi=150)
plt.show()

```
